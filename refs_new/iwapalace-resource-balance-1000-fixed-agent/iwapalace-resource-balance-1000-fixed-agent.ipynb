## Package contents / 同梱ファイル

The original ZIP is embedded in this notebook and restored by the first code cell, so the notebook can run even when Kaggle does not place auxiliary upload files next to the notebook source.

元の ZIP は Notebook 内に埋め込んであり、最初のコードセルで復元します。Kaggle 側で補助アップロードファイルが Notebook 横に配置されない場合でも実行できます。

- `submission_iwapalace_resource_balance_1000_fixed.zip`: final submission archive / 最終提出 ZIP
- `submission_files/main.py`: Kaggle entry point / Kaggle 提出エントリ
- `submission_files/agent.py`: self-contained heuristic + tabular Q policy / ヒューリスティックと Q モデルを統合した本体
- `submission_files/q_model_submit.json`: sanitized Q overlay / サニタイズ済み Q 値
- `submission_files/deck.csv`, `submission_files/deck_generated.csv`: deck files / デッキ CSV



```python
from pathlib import Path
from zipfile import ZipFile
import base64
import shutil

ZIP_B64 = ''
zip_path = Path('submission_iwapalace_resource_balance_1000_fixed.zip')
zip_path.write_bytes(base64.b64decode(ZIP_B64))

extract_dir = Path('submission_files')
if extract_dir.exists():
    shutil.rmtree(extract_dir)
extract_dir.mkdir()
with ZipFile(zip_path) as zf:
    zf.extractall(extract_dir)

print(zip_path, zip_path.stat().st_size, 'bytes')
for path in sorted(extract_dir.glob('*')):
    print(f'{path.name}: {path.stat().st_size} bytes')
```

## Rebuild the ZIP / ZIP 再生成

This recreates the same submission-file layout from `submission_files`. It is useful if you want to inspect or re-download the archive from the notebook output.

`submission_files` から提出 ZIP と同じファイル構成を再生成します。Notebook 出力から中身を確認・再取得したい場合に使えます。


```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED

source_dir = Path('submission_files')
output_zip = Path('submission_iwapalace_resource_balance_1000_fixed_rebuilt.zip')

with ZipFile(output_zip, 'w', ZIP_DEFLATED) as zf:
    for file in sorted(source_dir.iterdir()):
        if file.is_file():
            zf.write(file, file.name)

print(output_zip, output_zip.stat().st_size, 'bytes')
```
